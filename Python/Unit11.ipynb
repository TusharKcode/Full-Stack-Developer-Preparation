```python
#Creating a Frequency Distribution Table using Python
#Step 1: Import Libraries
import pandas as pd
import matplotlib. pyplot as plt
#Step 2: Data Collection
#Let us create a list of numbers as our dataset.
data = [23, 45, 23, 67, 89, 45, 23, 56, 34, 67, 23]
#Step 3: Create a Pandas DataFrame
#Convert the dataset into a Pandas DataFrame for easier manipulation. 
df = pd.DataFrame(data, columns=['Values'])
#Step 4: Calculate Frequencies
#Use the Pandas `value_counts()` method to calculate the frequencies of each unique value in the DataFrame.
frequency_table = df['Values'].value_counts().reset_index()
frequency_table.columns = ['Value', 'Frequency']
#Step 5: Sort the Frequency Table (Optional)
#Sorting the frequency table in ascending or descending order 
frequency_table = frequency_table.sort_values(by='Value', ascending=True)
#Step 6: Display the Frequency Distribution Table
#Print or display the frequency distribution table.
print(frequency_table)
#Step 7: Visualise the Distribution (Optional)
#You can also create a bar chart or histogram to visualise the frequency distribution.
plt.bar(frequency_table['Value'], frequency_table['Frequency'])
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Frequency Distribution')
plt.show()
```

```python
#Example : Frequency table
#Method 2 – With the help of crosstab() function
import pandas as pd
df = pd.DataFrame({'Student_Grade': ['A','B','A','B','B', 'B', 'B', 'C', 'C', 'D'],
                   'Student_Age': [18, 25, 28, 19, 30, 20, 15, 18, 29, 17],
                   'Student_Gender': ['M','F', 'M', 'F', 'F', 'M', 'M', 'F', 'F', 'F']})
print("The Dataset is : ")
print(df)
print("\nFrequency Table for the Grade in the dataset : ")
print(pd.crosstab(index=df['Student_Grade'], columns='count'))
```

```python
import pandas as pd
from scipy.stats import kruskal

# Dummy dataset
data = {
    "SALARY": [25000, 28000, 30000, 32000, 35000, 40000, 45000, 50000, 100000],
    "RANK": ['a','b','c','d','e','f','g','h','i']
}
dataset = pd.DataFrame(data)

# Group salaries by rank
groups = [dataset[dataset["RANK"] == r]["SALARY"] for r in dataset["RANK"].unique()]

# Kruskal-Wallis test
#The Kruskal–Wallis test is a nonparametric statistical test used to compare three or more independent groups 
# when the data does not follow a normal distribution. 
stat, p = kruskal(*groups)
print("Kruskal-Wallis Test: stat=%.3f, p=%.3f" % (stat, p))
```

```python
# The Mann–Whitney U test (also called the Wilcoxon rank-sum test) 
# is a nonparametric statistical test used to compare two independent groups 
# when the data does not necessarily follow a normal distribution.
from scipy.stats import mannwhitneyu

group_a = dataset[dataset["RANK"] == "a"]["SALARY"]
group_b = dataset[dataset["RANK"] == "b"]["SALARY"]

stat, p = mannwhitneyu(group_a, group_b)
print("Mann-Whitney U Test: stat=%.3f, p=%.3f" % (stat, p))
```

```python
import statsmodels.api as sm

# Encode rank as numeric
dataset["RANK_NUM"] = dataset["RANK"].rank(method="dense").astype(int)

X = dataset["RANK_NUM"]
y = dataset["SALARY"]

X = sm.add_constant(X)  # add intercept
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
# A treemap is a visualization that displays hierarchical data using nested rectangles. 
# Each rectangle represents a category (or subcategory), and its size corresponds to a quantitative value
import matplotlib.pyplot as plt
import squarify  # pip install squarify

# Dummy data: department budgets
departments = ["HR", "Finance", "IT", "Marketing", "Sales", "R&D"]
budgets = [15, 25, 30, 10, 40, 20]  # values in millions

# Create treemap
plt.figure(figsize=(10,6))
squarify.plot(
    sizes=budgets, 
    label=departments, 
    alpha=0.8, 
    color=["#FF9999","#66B2FF","#99FF99","#FFCC99","#FFD700","#C71585"]
)

plt.title("Company Department Budgets (Treemap)", fontsize=16)
plt.axis("off")  # hide axes
plt.show()
```
